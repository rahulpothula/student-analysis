```python
df["grade"] = np.select(
    [df["average"] >= 80, df["average"] >= 65, df["average"] >= 50],
    ["A", "B", "C"],
    default="F",
)
df["attendance_level"] = pd.cut(
    df["attendance_pct"], bins=[0, 75, 90, 100], labels=["Low", "Medium", "High"]
)
df[["average", "grade", "attendance_pct", "attendance_level"]].head()
```

```python
np.mean(df["average"])
np.median(df["average"])
np.std(df["average"])
np.percentile(df["average"], 90)
((df["passed"].mean() * 100))
```

```python
sns.histplot(df["average"], bins=20, kde=True, color="steelblue")
```

```python
plt.axvline(50, color="red", linestyle="--", label="Pass mark (50)")
plt.title("Distribution of Average Scores")
plt.xlabel("Average score")
plt.ylabel("Number of students")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# chart 1: study hours vs score
sns.regplot(data=df, x="study_hours_per_day", y="average")
plt.title("Study Hours vs Average Score")
plt.show()
```

```python
# chart 2: score by gender
sns.boxplot(data=df, x="gender", y="average")
plt.title("Average Score by Gender")
plt.show()
```

```python
# chart 3: correlation heatmap
cols = ["study_hours_per_day", "attendance_pct", "math", "science", "english", "average"]
sns.heatmap(df[cols].corr(), annot=True, cmap="coolwarm")
plt.show()
```

```python
# chart 4: score distribution
sns.histplot(df["average"], bins=20, kde=True)
plt.title("Distribution of Average Scores")
plt.show()
```

```python
# chart 5: average score by grade level
sns.barplot(data=df, x="grade_level", y="average")
plt.title("Average Score by Grade Level")
plt.show()
```
